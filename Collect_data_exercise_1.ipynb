```python
page=requests.get('https://ds.study.iitm.ac.in')
page.content
```

```python
soup = BeautifulSoup(page.content, 'html.parser')
print(soup.prettify())
```

```python
# Common attributes that might contain URLs
url_attributes = ['href', 'src', 'action', 'data-url']

all_links = set() # Use a set to store unique URLs

# Find all tags that might contain URLs in their attributes
for attr in url_attributes:
    elements_with_attr = soup.find_all(attrs={attr: True})
    for element in elements_with_attr:
        link = element.get(attr)
        if link is not None and link.strip() != '' and not link.startswith('#'):
            all_links.add(link)

# Print all extracted unique links
for link in sorted(list(all_links)):
    print(link)
```

Let's go through the code line by line to understand how it extracts links from the webpage:

1.  `url_attributes = ['href', 'src', 'action', 'data-url']`:
    *   This line initializes a list named `url_attributes`. This list contains common HTML attributes where URLs or links are typically stored.
    *   `'href'`: This is the most common attribute for hyperlinks, found in `<a>` tags.
    *   `'src'`: This attribute is used for sources of embedded content like images (`<img>`), scripts (`<script>`), iframes (`<iframe>`), and videos (`<video>`).
    *   `'action'`: This attribute is typically found in `<form>` tags and specifies where to send the form data when a form is submitted.
    *   `'data-url'`: This is a custom data attribute that developers might use to store URLs, especially in JavaScript-driven applications.

2.  `all_links = set() # Use a set to store unique URLs`:
    *   This line initializes an empty `set` named `all_links`. A set is a collection data type in Python that stores only unique elements. Using a set here is crucial because it automatically handles duplicate URLs, ensuring that each extracted link is stored only once.

3.  `for attr in url_attributes:`:
    *   This starts an outer loop that iterates through each attribute name (e.g., `'href'`, `'src'`) in the `url_attributes` list.

4.  `elements_with_attr = soup.find_all(attrs={attr: True})`:
    *   Inside the outer loop, this line uses `BeautifulSoup`'s `find_all()` method.
    *   `attrs={attr: True}` is a dictionary that tells `find_all()` to look for *any* HTML tag that has the current `attr` (e.g., `'href'`) defined, regardless of its value. It effectively finds all elements that possess this attribute.
    *   The result is a list of all `BeautifulSoup` tag objects that have the specified attribute.

5.  `for element in elements_with_attr:`:
    *   This starts an inner loop that iterates through each HTML `element` found in the `elements_with_attr` list for the current attribute.

6.  `link = element.get(attr)`:
    *   For each `element`, this line retrieves the value of the attribute specified by `attr` (e.g., `element.get('href')`).
    *   The `.get()` method is used here to safely retrieve the attribute's value. If the attribute doesn't exist on a particular element (even though `find_all` should have filtered for it), `.get()` will return `None` instead of raising an error.

7.  `if link is not None and link.strip() != '' and not link.startswith('#'):`:
    *   This is a conditional statement that checks if the extracted `link` is a valid and useful URL:
        *   `link is not None`: Ensures that an attribute value was actually found (i.e., it's not `None`).
        *   `link.strip() != ''`: Removes any leading/trailing whitespace from the link string and checks if it's not an empty string. This helps in filtering out attributes that might be present but have no value.
        *   `not link.startswith('#')`: This is an important filter. It excludes internal page anchors (like `#section1`) which are not external links to be crawled or considered unique URLs in most scraping contexts.

8.  `all_links.add(link)`:
    *   If the `link` passes all the validation checks, it is added to the `all_links` set. Because `all_links` is a set, if the same URL is encountered multiple times, it will only be stored once.

9.  `for link in sorted(list(all_links)):`:
    *   After iterating through all attributes and all relevant elements, this line prepares to print the collected unique links.
    *   `list(all_links)`: Converts the `set` of unique links into a `list`. This is necessary because sets are unordered, and we often want a consistent output order.
    *   `sorted(...)`: Sorts the list of links alphabetically. This makes the output easier to read and review.

10. `print(link)`:
    *   This line prints each unique, sorted link to the console.
